A description of the data is here:

*   *first*: When the flow (or this chunk of it) started, as Unix epoch seconds with millisecond precision.
*   *last*: When the last packet of the flow or chunk was seen, in the same format.
*  *duration*: last − first, in seconds. It's 0 for single-packet flows like DNS, NTP and scan probes.
* *proto*: IP protocol: TCP or UDP. In the v5 record these are the numbers 6 and 17.
* *src, sport*: Source IP address and port.
* *dst, dport*: Destination IP address and port.
* *tcp_flags*: Every TCP flag seen in the flow, combined. It's shown as six positions UAPRSF (URG, ACK, PSH, RST, SYN, FIN), with a dot for flags not seen. It's blank for UDP.
* *tos*: The IP Type of Service byte in decimal. It's almost always 0. The value 104 means DSCP 26, which marks RoCE traffic.
* *packets*: Number of packets in this record.
* *bytes*: Number of bytes in this record, counted at layer 3 (IP header plus payload, no Ethernet framing), as v5 does.
